```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("dylanjcastillo/7k-books-with-metadata")

print("Path to dataset files:", path)
```

```python
import pandas as pd
```

```python
book = pd.read_csv(f"{path}/books.csv")
```

```python
book
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
ax = plt.axes()
sns.heatmap(book.isna().transpose(), cbar=False ,ax =ax)


plt.xlabel("Columns")
plt.ylabel("Missing Values")

plt.show()
```

```python
import numpy as np

book["missing_description"] = np.where(book["description"].isna(),1,0)
book["age_of_the_book"] = 2025 - book["published_year"]
```

```python
columns_of_intrest =["missing_description", "age_of_the_book","num_pages","average_rating"]

correlation_matrix = book[columns_of_intrest].corr(method="spearman")

sns.set_theme(style ="white")
plt.figure(figsize=(8,6))

heatmap =sns.heatmap(correlation_matrix, annot=True,fmt=".2f",cmap="coolwarm", cbar_kws={"label": "Spearman Correlation"})

plt.title("Correlation Heatmap")
plt.show()
```

```python
book[(book["description"].isna()) |
     (book["num_pages"].isna()) |
     (book["average_rating"].isna()) |
     (book["published_year"].isna()) ]
```

```python
book_missing =book[(~book["description"].isna()) &
     (~book["num_pages"].isna()) &
     (~book["average_rating"].isna()) &
     (~book["published_year"].isna()) ].copy()

book_missing
```

```python
book_missing["categories"].value_counts().reset_index().sort_values("count", ascending=False)
```

```python
book_missing["word_in_description"] = book_missing["description"].str.split().str.len()


book_missing
```

```python
book_missing_25 = book_missing[book_missing["word_in_description"] >= 25].copy()

book_missing_25
```

```python
book_missing_25["title_and_subtitle"] = (np.where(book_missing_25["subtitle"].isna(),book_missing_25["title"],
                                         book_missing_25[["title","subtitle"]].astype(str).agg(":".join, axis=1)))

book_missing_25
```

```python
book_missing_25["tagged_description"] = book_missing_25[["isbn13","description"]].astype(str).agg(" ".join, axis=1)

book_missing_25
```

```python
(

    book_missing_25.drop(["subtitle","missing_description","age_of_the_book","word_in_description"], axis=1)
    .to_csv("book_cleaned.csv", index=False)

)
```
